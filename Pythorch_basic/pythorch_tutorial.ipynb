1. Introduction

Definition:
PyTorch is an open-source deep learning framework developed by Facebook AI Research (FAIR) used for building and training neural networks. It provides dynamic computation graphs, which allow flexibility for research and debugging.

Key Points for Interview:

Pythonic and easy to debug.

Supports GPU acceleration using CUDA.

Used widely in research and production.

Main alternatives: TensorFlow, JAX.

Dynamic vs Static Graphs:

Dynamic Graph (PyTorch): Built on-the-fly during execution. Good for flexible architectures (like RNNs).

Static Graph (TensorFlow 1.x): Defined before execution, optimized but less flexible

2. PyTorch Basics: Tensors

Definition:

A Tensor is a multi-dimensional array similar to NumPy arrays, but can run on GPU for faster computation. Tensors are the building blocks of PyTorch.

Types of Tensors:

1D Tensor → Vector

2D Tensor → Matrix

3D Tensor or higher → For images, videos, etc.

Key Operations:

Creation: torch.tensor(), torch.zeros(), torch.ones(), torch.rand()

GPU support: .to('cuda') or .cuda()

```python
# 🔹 3️⃣ Concatenation & Stacking
# Concatenation (torch.cat):

# Joins tensors along an existing dimension.
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6]])
cat = torch.cat((a, b), dim=0)
print(cat)
print(cat.shape)
```

```python
# Stacking (torch.stack)
# Stacking combines tensors along a new dimension, unlike torch.cat which joins along an existing dimension.
# Adds a new dimension while combining tensors.
x = torch.tensor([1, 2])
y = torch.tensor([3, 4])
stacked = torch.stack((x, y))
print(stacked)
print(stacked.shape)
```

```python
import torch

a = torch.tensor([1, 2])
b = torch.tensor([3, 4])
c = torch.tensor([5, 6])

stack0 = torch.stack((a, b, c), dim=0)  # stack along new first dimension
stack1 = torch.stack((a, b, c), dim=1)  # stack along new second dimension

print("dim=0:", stack0)
print("dim=0 shape:", stack0.shape)
print("--------------")
print("dim=1:", stack1)
print("dim=1 shape:", stack1.shape)
```

```python
# unsqueeze(dim) → adds a new dimension

# cat() → concatenates along that dimension

# Output is identical to torch.stack((x, y, z), dim=0)

# This helps understand stacking as a combination of unsqueeze + concat.
x = torch.tensor([1,2])
y = torch.tensor([3,4])
z = torch.tensor([5,6])

stack_manual = torch.cat((x.unsqueeze(0), y.unsqueeze(0), z.unsqueeze(0)), dim=0)
print(stack_manual)
```

```python
# torch.hstack and torch.vstack (specialized stacking)

# torch.vstack → stack along rows (vertical, dim=0 for 2D)

# torch.hstack → stack along columns (horizontal, dim=1 for 2D)


a = torch.tensor([[1,2]])
b = torch.tensor([[3,4]])

v = torch.vstack((a,b))
h = torch.hstack((a,b))

print("vstack:", v)
print("hstack:", h)
```

```python
# Cloning vs Detaching

# clone()

# Creates a copy of tensor with the gradient tracking.

x = torch.tensor([1,2,3.], requires_grad=True)
y = x.clone()
print(y.requires_grad)  # True
y
```

```python
# detach()

# Creates a new tensor without tracking gradients (used in inference).✅ Use .detach() when you want a tensor but don’t want autograd to follow it.

y = x.detach()
print(y.requires_grad)  # False
```

```python
# Type Casting & Device Movement
# Data type conversion
x = torch.tensor([1, 2, 3])
print(x.float())   # Convert to float32
print(x.double())  # Convert to float64
print(x.int())     # Convert to int32
```

```python
# eduction Operations (Summarizing Data)
# Sum, Mean, Max, etc.

# Use dim to control which axis to reduce:

# dim=0 → column-wise

# dim=1 → row-wise
x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
x=x.float()

print(x.sum())       # All elements
print(x.mean(dim=0)) # Column-wise mean
print(x.max(dim=1))  # Row-wise max + indices
```

```python
# Comparison Operations
x = torch.tensor([1, 2, 3])
y = torch.tensor([2, 2, 2])

print(x == y)   # element-wise equality
print(x > y)    # greater than
print(torch.equal(x, y))  # all elements equal?
```

```python
# squeeze()

# Definition:
# Removes dimensions of size 1 from a tensor.

# Purpose:

# Helps simplify tensor shapes.

# Common in CNNs to remove extra singleton dimensions after convolution or pooling.


x = torch.randn(1, 3, 1, 4)
y = x.squeeze()  # removes all dimensions with size 1
print(y.shape)   # torch.Size([3, 4])

# Remove specific dimension
z = x.squeeze(0)  # remove only dim 0
print(z.shape)    # torch.Size([3, 1, 4])
```

```python
# unsqueeze()

# Definition:
# Adds a dimension of size 1 at the specified position.     squeeze and unsqueeze are commonly used together to align shapes for neural networks.

# Purpose:

# Useful to match input shapes for broadcasting.

# Often used in CNNs/RNNs to add batch or channel dimensions.


x = torch.tensor([1,2,3])       # shape (3)
y = x.unsqueeze(0)               # add batch dim at 0 → (1,3)
z = x.unsqueeze(1)               # add dim at 1 → (3,1)
```

```python
# permute()

# Definition:
# Reorders dimensions of a tensor according to the given order.

# Purpose:

# Often used to convert between tensor layouts, e.g., (H, W, C) ↔ (C, H, W) for image processing.

# Does not change data, just changes how dimensions are viewed.


x = torch.randn(2, 3, 4)   # shape (2,3,4)
y = x.permute(1, 0, 2)     # shape (3,2,4)
```

```python
# Rank / Order

# Definition: Number of dimensions a tensor has.
# Purpose: Identifies if it’s a scalar (0D), vector (1D), matrix (2D), or higher.

# Example:

scalar = torch.tensor(5)          # 0D
vector = torch.tensor([1, 2, 3])  # 1D
matrix = torch.tensor([[1,2],[3,4]]) # 2D
print(scalar.ndim)  # Output: 0 1 2
print(vector.ndim)
print(matrix.ndim)
print(scalar.shape,vector.shape,matrix.shape)

```

```python
# Static vs Dynamic Computation Graphs

# Static Graphs (e.g., TensorFlow 1.x):
# You must define the entire computation before running it (like a fixed blueprint).

# Dynamic Graphs (PyTorch):
# The graph is built on the fly as operations are executed — perfect for debugging and variable-length inputs.


# derivative
x = torch.tensor(2.0, requires_grad=True)
y = x**2 + 3*x + 2
y.backward()
print(x.grad)  # derivative = 2x + 3 → 2*2 + 3 = 7
```
